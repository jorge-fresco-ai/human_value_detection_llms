```python
agente_openess = Agent(model=ollama_model)
agente_subdimensiones_openess = Agent(model=ollama_model)
agente_enhancement = Agent(model=ollama_model)
agente_subdimensiones_enhancement = Agent(model=ollama_model)
agente_conservation = Agent(model=ollama_model)
agente_subdimensiones_conservation = Agent(model=ollama_model)
agente_transcendence = Agent(model=ollama_model)
agente_subdimensiones_transcendence = Agent(model=ollama_model)
agente_polaridad = Agent(model=ollama_model)
```

```python
import json
from typing import Dict, Any

def process_dimension(input_text: str, main_agent: Agent, sub_agent: Agent, 
                     polaridad_agent: Agent, dimension_name: str) -> Dict[str, Any]:
    """
    Procesa una dimensión específica (openness, enhancement, etc.)
    """
    main_prompt = f"""
    Analiza el siguiente texto para determinar si presenta características de {dimension_name}.
    Responde únicamente con 'true' o 'false'.
    
    Texto: {input_text}
    """
    
    main_result = main_agent(main_prompt)
    print(f"Main raw result for {dimension_name}: {main_result}")
    print('##########')
    print(main_result.message['content'][0]['text'])
    print(type(main_result))

    if 'true' in main_result.message['content'][0]['text'].lower():
        is_dimension_present = 'true'

    print(f"Is {dimension_name} present? {str(is_dimension_present)}")
    
    result = {
        dimension_name: is_dimension_present,
        "subdimensiones": {},
        "polaridad": False
    }
    
    if is_dimension_present:
        sub_prompt = f"""
        Analiza las subdimensiones de {dimension_name} en el siguiente texto.
        Evalúa cada una de las 7 subdimensiones y responde con un JSON con formato:
        {{"1": true/false, "2": true/false, "3": true/false, "4": true/false, 
          "5": true/false, "6": true/false, "7": true/false}}
        
        Texto: {input_text}
        """
        
        sub_result = sub_agent(sub_prompt)
        try:
            subdimensiones = json.loads(sub_result)
            result["subdimensiones"] = subdimensiones
        except json.JSONDecodeError:
            result["subdimensiones"] = {str(i): False for i in range(1, 8)}
        
        pol_prompt = f"""
        Analiza la polaridad de {dimension_name} en el siguiente texto.
        Responde únicamente con 'true' si es positiva o 'false' si es negativa.
        
        Texto: {input_text}
        """
        
        pol_result = polaridad_agent(pol_prompt)
        print(f"Polaridad raw result for {dimension_name}: {pol_result}")
        result["polaridad"] = pol_result == 'true'
    
    return result

def analyze_text_flow(input_text: str) -> Dict[str, Any]:
    """
    Flujo principal que procesa el texto a través de todas las dimensiones
    """
    final_result = {}
    
    dimensions = [
        ("openness", agente_openess, agente_subdimensiones_openess),
        ("enhancement", agente_enhancement, agente_subdimensiones_enhancement),
        ("conservation", agente_conservation, agente_subdimensiones_conservation),
        ("transcendence", agente_transcendence, agente_subdimensiones_transcendence)
    ]
    
    for dimension_name, main_agent, sub_agent in dimensions:
        print(f"Procesando dimensión: {dimension_name}")
        
        dimension_result = process_dimension(
            input_text, 
            main_agent, 
            sub_agent, 
            agente_polaridad,  
            dimension_name
        )
        
        final_result[dimension_name] = dimension_result
    
    return final_result

def run_analysis(text: str) -> str:
    """
    Ejecuta el análisis completo y devuelve el JSON resultado
    """
    try:
        result = analyze_text_flow(text)
        return json.dumps(result, indent=2, ensure_ascii=False)
    except Exception as e:
        return json.dumps({"error": str(e)}, indent=2)

# Ejemplo de uso
if __name__ == "__main__":
    input_text = "I'm totally open to new experiences and love exploring different cultures and ideas."
    
    print("Iniciando análisis...")
    resultado = run_analysis(input_text)
    print("Resultado:")
    print(resultado)
```

```python
import json
import yaml
from typing import Dict, Any
from pathlib import Path


class PromptManager:
    """Maneja la carga y acceso a los prompts desde YAML"""
    
    def __init__(self, config_path: str = "/mnt/c/Users/BorjaEsteveMolner/Documents/human_value_detection_llms/prompts.yaml"):
        self.config_path = config_path
        self.prompts = self._load_prompts()
    
    def _load_prompts(self) -> Dict[str, Any]:
        """Carga los prompts desde el archivo YAML"""
        try:
            with open(self.config_path, 'r', encoding='utf-8') as file:
                config = yaml.safe_load(file)
                return config['agents']
        except FileNotFoundError:
            raise FileNotFoundError(f"No se encontró el archivo de configuración: {self.config_path}")
        except yaml.YAMLError as e:
            raise ValueError(f"Error al parsear el archivo YAML: {e}")
    
    def get_agent_prompts(self, agent_name: str) -> Dict[str, str]:
        """Obtiene los prompts de un agente específico"""
        if agent_name not in self.prompts:
            raise KeyError(f"Agente '{agent_name}' no encontrado en la configuración")
        return self.prompts[agent_name]
    
    def format_user_prompt(self, agent_name: str, **kwargs) -> str:
        """Formatea el user_prompt con los parámetros dados"""
        prompts = self.get_agent_prompts(agent_name)
        return prompts['user_prompt'].format(**kwargs)
    
    def get_system_prompt(self, agent_name: str) -> str:
        """Obtiene el system_prompt de un agente"""
        prompts = self.get_agent_prompts(agent_name)
        return prompts['system_prompt']


class AgentManager:
    """Maneja la creación y configuración de agentes"""
    
    def __init__(self, model, prompt_manager: PromptManager):
        self.model = model
        self.prompt_manager = prompt_manager
        self.agents = {}
        self._initialize_agents()
    
    def _initialize_agents(self):
        """Inicializa todos los agentes con sus prompts correspondientes"""
        agent_names = [
            "agente_openness",
            "agente_subdimensiones_openness", 
            "agente_enhancement",
            "agente_subdimensiones_enhancement",
            "agente_conservation",
            "agente_subdimensiones_conservation",
            "agente_transcendence",
            "agente_subdimensiones_transcendence",
            "agente_polaridad"
        ]
        
        for agent_name in agent_names:
            try:
                system_prompt = self.prompt_manager.get_system_prompt(agent_name)
                agent = Agent(model=self.model, system_prompt=system_prompt)
                self.agents[agent_name] = agent
                print(f"✓ Agente '{agent_name}' inicializado correctamente")
            except KeyError:
                print(f"⚠ Advertencia: No se encontraron prompts para '{agent_name}'")
    
    def get_agent(self, agent_name: str):
        """Obtiene un agente específico"""
        if agent_name not in self.agents:
            raise KeyError(f"Agente '{agent_name}' no está disponible")
        return self.agents[agent_name]


class TextAnalyzer:
    """Clase principal para el análisis de texto"""
    
    def __init__(self, model, config_path: str = "/mnt/c/Users/BorjaEsteveMolner/Documents/human_value_detection_llms/prompts.yaml"):
        self.prompt_manager = PromptManager(config_path)
        self.agent_manager = AgentManager(model, self.prompt_manager)
    
    def process_dimension(self, input_text: str, dimension_name: str) -> Dict[str, Any]:
        """Procesa una dimensión específica (openness, enhancement, etc.)"""
        
        agent_mapping = {
            "openness": ("agente_openness", "agente_subdimensiones_openness"),
            "enhancement": ("agente_enhancement", "agente_subdimensiones_enhancement"),
            "conservation": ("agente_conservation", "agente_subdimensiones_conservation"),
            "transcendence": ("agente_transcendence", "agente_subdimensiones_transcendence")
        }
        
        if dimension_name not in agent_mapping:
            raise ValueError(f"Dimensión '{dimension_name}' no reconocida")
        
        main_agent_name, sub_agent_name = agent_mapping[dimension_name]
        
        main_agent = self.agent_manager.get_agent(main_agent_name)
        sub_agent = self.agent_manager.get_agent(sub_agent_name)
        polaridad_agent = self.agent_manager.get_agent("agente_polaridad")
        
        main_prompt = self.prompt_manager.format_user_prompt(main_agent_name, input_text=input_text)
        main_result = main_agent(main_prompt)
        
        print(f"Main raw result for {dimension_name}: {main_result}")
        print('##########')
        print(main_result.message['content'][0]['text'])
        
        is_dimension_present = 'true' in main_result.message['content'][0]['text'].lower()
        print(f"Is {dimension_name} present? {is_dimension_present}")
        
        result = {
            dimension_name: is_dimension_present,
            "subdimensiones": {},
            "polaridad": False
        }
        
        if is_dimension_present:
            sub_prompt = self.prompt_manager.format_user_prompt(sub_agent_name, input_text=input_text)
            sub_result = sub_agent(sub_prompt)
            
            try:
                subdimensiones_text = sub_result.message['content'][0]['text']
                subdimensiones = json.loads(subdimensiones_text)
                result["subdimensiones"] = subdimensiones
            except (json.JSONDecodeError, KeyError, IndexError) as e:
                print(f"Error procesando subdimensiones: {e}")
                result["subdimensiones"] = {str(i): False for i in range(1, 8)}
            
            pol_prompt = self.prompt_manager.format_user_prompt(
                "agente_polaridad", 
                input_text=input_text, 
                dimension_name=dimension_name
            )
            pol_result = polaridad_agent(pol_prompt)
            print(f"Polaridad raw result for {dimension_name}: {pol_result}")
            
            try:
                polaridad_text = pol_result.message['content'][0]['text'].lower()
                result["polaridad"] = 'true' in polaridad_text
            except (KeyError, IndexError) as e:
                print(f"Error procesando polaridad: {e}")
                result["polaridad"] = False
        
        return result
    
    def analyze_text_flow(self, input_text: str) -> Dict[str, Any]:
        """Flujo principal que procesa el texto a través de todas las dimensiones"""
        final_result = {}
        
        dimensions = ["openness", "enhancement", "conservation", "transcendence"]
        
        for dimension_name in dimensions:
            print(f"Procesando dimensión: {dimension_name}")
            
            try:
                dimension_result = self.process_dimension(input_text, dimension_name)
                final_result[dimension_name] = dimension_result
            except Exception as e:
                print(f"Error procesando dimensión {dimension_name}: {e}")
                final_result[dimension_name] = {
                    dimension_name: False,
                    "subdimensiones": {str(i): False for i in range(1, 8)},
                    "polaridad": False,
                    "error": str(e)
                }
        
        return final_result
    
    def run_analysis(self, text: str) -> str:
        """Ejecuta el análisis completo y devuelve el JSON resultado"""
        try:
            result = self.analyze_text_flow(text)
            return json.dumps(result, indent=2, ensure_ascii=False)
        except Exception as e:
            return json.dumps({"error": str(e)}, indent=2)

# TODO[]: Refactorizar el código para mejorar la legibilidad y modularidad.
# TODO[]: Añadir manejo de errores más robusto.
# TODO[]: Testear bien lo de los true/false, se podría mejorar
# TODO[]: Añadir que se pueda configurar el modelo desde fuera (cada agente su modelo)
# TODO[]: Sustituir prints por logging y niveles (info/debug/warning).
# TODO[]: Permitir pasar la lista de agentes por config en vez de estar fija en el código.

if __name__ == "__main__":

    input_text = "I'm totally open to new experiences and love exploring different cultures and ideas."
    
    print("Iniciando análisis...")
    
    analyzer = TextAnalyzer(ollama_model)
    resultado = analyzer.run_analysis(input_text)
    
    print("Resultado:")
    print(resultado)
```
